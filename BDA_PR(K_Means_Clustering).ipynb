# **K-Means Clustering using Python Scikit-Learn**

```python
# Step 1: Import libraries

import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Step 2: Create dataset

X = [
    [1,2],[2,1],[2,3],
    [8,8],[9,8],[8,9],
    [15,2],[16,3],[15,4]
]
```

```python
# Step 3: Create K-Means model

model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)
```

```python
# Step 4: Aplly K-Means clustering

model.fit(X)
```

```python
# Step 5: Display cluster labels

print("Cluster Labels:")
print(model.labels_)
```

```python
# Step 6: Display centroid

print("\nCluster Centroid:")
print(model.cluster_centers_)
```

```python
# Step 7: Visualize clusters

plt.scatter(
    [p[0] for p in X],
    [p[1] for p in X],
    c=model.labels_

)

plt.scatter(
    model.cluster_centers_[:,0],
    model.cluster_centers_[:,1],
    marker="X",
    s=200,
    label="Centroids"
)

plt.title("K-Means Clustering")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.show()
```
